```python
import pandas as pd

# Row Selection (filtering)
# Select rows where petal_length > 4.5 AND species is "Iris-virginica"
# Note: '&' is used for 'AND', and conditions are wrapped in parentheses
if 'iris_df' not in globals():
    iris_df = pd.read_csv("IRIS.csv")

filtered_iris_df = iris_df[(iris_df['petal_length'] > 4.5) & (iris_df['species'] == 'Iris-virginica')]

print("\n--- Filtered Rows (Petal Length > 4.5 & Species is Iris-virginica) ---")
print(filtered_iris_df)
```

```python
# Grouping and Aggregation
# Group by species & calculate specific metric of each group 

grouped_summary = iris_df.groupby('species').agg(
    average_sepal_length=('sepal_length', 'mean'),
    max_petal_width=('petal_width', 'max'),
    std_sepal_width=('sepal_width', 'std')
)

print("\n--- Grouped Summary by Species ---")
print(grouped_summary)
```

```python
# Create New Column and Find average

iris_df['petal_ratio'] = iris_df['petal_length'] / iris_df['petal_width'] # Column - Petal ratio
avg_petal_ratio = iris_df.groupby('species')['petal_ratio'].mean() # Average petal ratio by species

print("\n--- Average Petal Ratio by Species ---")
print(avg_petal_ratio)
```

# TItanic Dataset

```python
df = pd.read_csv('Titanic-Dataset.csv')
```

```python
# Display specific columns (e.g., 'Name', 'Age', 'Survived')

selected_columns = df[['Name', 'Sex', 'Age', 'Fare', 'Survived']]
print("--- Selected Columns ---")
print(selected_columns.head())  # .head() limits output readability
```

```python
# Passenger Selection (filtering)
filtered_passengers = df[(df['Sex'] == 'female') & (df['Fare'] > 30)]

print("\n--- Female Passengers with Fare > 30 ---")
print(filtered_passengers)
```

```python
# Grouping and Aggregation by Pclass
pclass_summary = df.groupby('Pclass').agg(
    survival_rate=('Survived', 'mean'),
    average_fare=('Fare', 'mean'),
    average_age=('Age', 'mean')
)

print("\n--- Summary Statistics by Passenger Class (Pclass) ---")
print(pclass_summary)
```
